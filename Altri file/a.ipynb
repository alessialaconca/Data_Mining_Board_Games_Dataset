```python
# Loaded variable 'df' from URI: /Users/ale/Documents/primo anno/datamining/dm1_25_26_dataset/DM1_game_dataset.csv
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv(r'/Users/ale/Documents/primo anno/datamining/dm1_25_26_dataset/DM1_game_dataset.csv')

#Parentesi Quadre come null
df = df.replace("[]", np.nan)

#Funzione di dispersione
def dispersione(Disp):
    return pd.Series({
        "Varianza": Disp.var(),
        "Deviazione std": Disp.std(),
        "Range": Disp.max() - Disp.min(),
        "IQR": Disp.quantile(0.75) - Disp.quantile(0.25),
        "Coeff. Variazione": (Disp.std() / Disp.mean())
    })

```

```python
df = df.fillna({'ComAgeRec': 0})
print(df['ComAgeRec'].corr(df['MfgAgeRec']))
righe_con_zero0 = df[(df['ComAgeRec'] == 0)]
print(len(righe_con_zero0))
# Filtra il DataFrame per trovare le righe in cui entrambe le colonne sono 0
righe_con_zero = df[(df['ComAgeRec'] == 0) & (df['MfgAgeRec'] == 0)]

# Conta il numero di righe risultanti
conteggio = len(righe_con_zero)

# Stampa il risultatox
print(f"\nIl numero di righe in cui sia 'ComAgeRec' che 'MfgAgeRec' sono 0 è: {conteggio}")
```
